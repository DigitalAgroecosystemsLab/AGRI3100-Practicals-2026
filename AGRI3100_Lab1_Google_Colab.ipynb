## 2. Simple calculations


```python
field_area = 20
yield_per_ha = 4.2
total_yield = field_area * yield_per_ha

print("Field area:", field_area, "ha")
print("Yield:", yield_per_ha, "tonnes/ha")
print("Total estimated production:", total_yield, "tonnes")
```

## 3. Agricultural variables


```python
crop = "Wheat"
soil_moisture = 18.5
ndvi = 0.72
nitrogen_rate = 80
yield_t_ha = 4.1

print("Crop:", crop)
print("Soil moisture:", soil_moisture, "%")
print("NDVI:", ndvi)
print("Nitrogen rate:", nitrogen_rate, "kg/ha")
print("Yield:", yield_t_ha, "t/ha")
```

## 4. Load Python libraries


```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries successfully loaded!")
```

## 5. Create a Digital Agriculture dataset


```python
data = {
    "Plot_ID": ["P01","P02","P03","P04","P05","P06","P07","P08","P09","P10","P11","P12"],
    "Crop": ["Wheat","Wheat","Canola","Canola","Wheat","Canola","Wheat","Canola","Wheat","Canola","Wheat","Canola"],
    "Nitrogen_kg_ha": [60,80,70,90,100,110,50,100,90,80,110,60],
    "Soil_Moisture_pct": [18.2,20.1,17.4,21.5,22.0,19.8,16.5,23.1,20.8,18.9,24.0,17.8],
    "NDVI": [0.54,0.62,0.58,0.71,0.76,0.67,0.49,0.79,0.73,0.61,0.82,0.52],
    "Yield_t_ha": [3.2,3.8,2.7,3.4,4.5,3.2,2.8,3.8,4.1,2.9,4.8,2.5]
}

df = pd.DataFrame(data)
df
```

## 6. Explore the dataset


```python
df.head()
```

```python
print("Dataset shape:", df.shape)
print("Number of agricultural plots:", df.shape[0])
print("Number of variables:", df.shape[1])
print("Variables:", df.columns.tolist())
```

## 7. Basic statistics


```python
df.describe()
```

```python
print("Average crop yield:", round(df["Yield_t_ha"].mean(), 2), "t/ha")
print("Average NDVI:", round(df["NDVI"].mean(), 2))
print("Minimum yield:", df["Yield_t_ha"].min(), "t/ha")
print("Maximum yield:", df["Yield_t_ha"].max(), "t/ha")
```

## 8. Agricultural questions


```python
highest_ndvi_plot = df.loc[df["NDVI"].idxmax()]
highest_ndvi_plot
```

```python
high_moisture = df[df["Soil_Moisture_pct"] > 20]
high_moisture
```

```python
high_ndvi = df[df["NDVI"] > 0.70]
high_ndvi
```

```python
selected_plots = df[(df["Crop"] == "Wheat") & (df["NDVI"] > 0.70)]
selected_plots
```

## 9. Compare crops


```python
df.groupby("Crop")["Yield_t_ha"].mean()
```

```python
df.groupby("Crop")[["Nitrogen_kg_ha","Soil_Moisture_pct","NDVI","Yield_t_ha"]].mean()
```

## 10. Visualize crop yield


```python
plt.figure(figsize=(10,5))
plt.bar(df["Plot_ID"], df["Yield_t_ha"])
plt.xlabel("Plot")
plt.ylabel("Yield (t/ha)")
plt.title("Crop Yield Across Agricultural Plots")
plt.tight_layout()
plt.show()
```

## 11. Soil moisture versus yield


```python
plt.figure(figsize=(7,5))
plt.scatter(df["Soil_Moisture_pct"], df["Yield_t_ha"], s=80)
plt.xlabel("Soil Moisture (%)")
plt.ylabel("Yield (t/ha)")
plt.title("Soil Moisture vs. Crop Yield")
plt.tight_layout()
plt.show()
```

## 12. NDVI versus yield


```python
plt.figure(figsize=(7,5))
plt.scatter(df["NDVI"], df["Yield_t_ha"], s=90)
plt.xlabel("NDVI")
plt.ylabel("Yield (t/ha)")
plt.title("Relationship Between NDVI and Crop Yield")
plt.tight_layout()
plt.show()
```

## 13. NDVI versus yield with plot labels


```python
plt.figure(figsize=(8,6))
plt.scatter(df["NDVI"], df["Yield_t_ha"], s=90)

for i in range(len(df)):
    plt.text(
        df["NDVI"].iloc[i] + 0.005,
        df["Yield_t_ha"].iloc[i],
        df["Plot_ID"].iloc[i]
    )

plt.xlabel("NDVI")
plt.ylabel("Yield (t/ha)")
plt.title("NDVI and Crop Yield by Plot")
plt.tight_layout()
plt.show()
```

## 14. Correlation between NDVI and yield


```python
correlation = df["NDVI"].corr(df["Yield_t_ha"])
print("Correlation between NDVI and yield:", round(correlation, 2))
```

A correlation close to:

- **+1** = strong positive relationship
- **0** = little or no linear relationship
- **‒1** = strong negative relationship


## 15. Turn data into information


```python
average_yield = df["Yield_t_ha"].mean()

df["Yield_Status"] = df["Yield_t_ha"].apply(
    lambda x: "Above Average" if x > average_yield else "Below Average"
)

df[["Plot_ID", "Yield_t_ha", "Yield_Status"]]
```

## 16. Simple management screening


```python
attention_plots = df[
    (df["NDVI"] < 0.60) |
    (df["Yield_t_ha"] < 3.0)
]

attention_plots
```

> **Important:** Python is not making the agronomic decision. It is helping us identify where we should investigate.
